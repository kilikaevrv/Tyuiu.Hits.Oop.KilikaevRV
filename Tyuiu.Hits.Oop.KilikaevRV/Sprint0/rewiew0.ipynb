```
using System;

class Plant
{
    public string Name;
    public string Type;
    public float Height;

    public virtual void Bloom()
    {
        Console.WriteLine("Растение цветет");
    }

    public void Grow(float amount)
    {
        Height = Height + amount;
        Console.WriteLine("Высота растения: " + Height);
    }
}

class Flower : Plant
{
    public string Color;
    public int PetalCount;

    public override void Bloom()
    {
        Console.WriteLine("Цветок " + Name + " распускается");
    }

    public void Wither()
    {
        Console.WriteLine("Цветок " + Name + " увядает");
    }
}

class Tree : Plant
{
    public int Age;
    public string LeafType;

    public override void Bloom()
    {
        Console.WriteLine("Дерево " + Name + " покрывается листьями");
    }

    public void GrowOlder()
    {
        Age = Age + 1;
        Console.WriteLine("Возраст дерева: " + Age);
    }
}

class Program
{
    public static void Main()
    {
        Flower flower = new Flower();

        flower.Name = "Роза";
        flower.Type = "Цветок";
        flower.Height = 35;
        flower.Color = "Красный";
        flower.PetalCount = 25;

        Tree tree = new Tree();

        tree.Name = "Береза";
        tree.Type = "Дерево";
        tree.Height = 500;
        tree.Age = 20;
        tree.LeafType = "Лиственные";

        Console.WriteLine("ЦВЕТОК");

        flower.Bloom();
        flower.Grow(5);
        flower.Wither();

        Console.WriteLine();

        Console.WriteLine("ДЕРЕВО");

        tree.Bloom();
        tree.Grow(30);
        tree.GrowOlder();

        Console.WriteLine();

        Console.WriteLine("ПОЛИМОРФИЗМ");

        Plant plant1 = flower;
        Plant plant2 = tree;

        plant1.Bloom();
        plant2.Bloom();
    }
}

Program.Main();
```
