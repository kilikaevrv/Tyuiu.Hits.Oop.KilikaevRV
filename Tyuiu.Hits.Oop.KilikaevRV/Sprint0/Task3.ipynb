<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Order в C#, который будет представлять информацию о
заказах товаров или услуг. На основе этого класса разработать 2-3 производных
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из
классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма.
Требования к базовому классу Order:
• Атрибуты: ID заказа (OrderId), Дата создания (CreationDate), Сумма заказа
(TotalAmount).
• Методы:
o CalculateTotal(): метод для расчета общей суммы заказа.
o AddItem(Item item): метод для добавления элемента в заказ.
o RemoveItem(Item item): метод для удаления элемента из заказа.


<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;
using System.Collections.Generic;

public class Item
{
    public string Name;
    public double Price;

    public Item(string name, double price)
    {
        Name = name;
        Price = price;
    }
}

public class Order
{
    public int OrderId;
    public DateTime CreationDate;
    public double TotalAmount;

    protected List<Item> items = new List<Item>();

    public Order(int id)
    {
        OrderId = id;
        CreationDate = DateTime.Now;
    }

    public virtual void CalculateTotal()
    {
        TotalAmount = 0;

        foreach (Item item in items)
            TotalAmount += item.Price;

        Console.WriteLine("Сумма заказа: " + TotalAmount);
    }

    public virtual void AddItem(Item item)
    {
        items.Add(item);
        Console.WriteLine("Добавлен товар: " + item.Name);
    }

    public virtual void RemoveItem(Item item)
    {
        items.Remove(item);
        Console.WriteLine("Удален товар: " + item.Name);
    }
}

public class OnlineOrder : Order
{
    public string CustomerEmail;

    public OnlineOrder(int id, string email) : base(id)
    {
        CustomerEmail = email;
    }

    public override void AddItem(Item item)
    {
        base.AddItem(item);
        Console.WriteLine("Способ доставки: курьерская доставка");
    }
}

public class PhysicalOrder : Order
{
    public string DeliveryAddress;

    public PhysicalOrder(int id, string address) : base(id)
    {
        DeliveryAddress = address;
    }

    public override void RemoveItem(Item item)
    {
        base.RemoveItem(item);
        Console.WriteLine("Товар возвращен");
    }
}

public class SpecializedOrder : Order
{
    public string SpecialConditions;

    public SpecializedOrder(int id, string conditions) : base(id)
    {
        SpecialConditions = conditions;
    }

    public override void CalculateTotal()
    {
        base.CalculateTotal();

        if (SpecialConditions == "Скидка 10%")
            TotalAmount *= 0.9;

        Console.WriteLine("Итоговая сумма: " + TotalAmount);
    }
}


// Проверка работы программы

Item item1 = new Item("Ноутбук", 50000);
Item item2 = new Item("Мышь", 2000);

OnlineOrder onlineOrder = new OnlineOrder(1, "user@mail.ru");
onlineOrder.AddItem(item1);
onlineOrder.CalculateTotal();

Console.WriteLine();

PhysicalOrder physicalOrder = new PhysicalOrder(2, "ул. Ленина, 10");
physicalOrder.AddItem(item2);
physicalOrder.RemoveItem(item2);

Console.WriteLine();

SpecializedOrder specializedOrder = new SpecializedOrder(3, "Скидка 10%");
specializedOrder.AddItem(item1);
specializedOrder.CalculateTotal();
```
