```
using System;

// Интерфейс (Контракт)
public interface IOrderManager
{
    void ProcessOrder();
    void CancelOrder();
}

// Базовый класс
public abstract class Order
{
    private decimal _totalAmount;
    
    public int OrderId { get; set; }
    public DateTime CreationDate { get; set; }
    public string CustomerName { get; set; }
    
    public decimal TotalAmount
    {
        get => _totalAmount;
        set
        {
            if (value < 0) throw new ArgumentException("Сумма не может быть отрицательной");
            _totalAmount = value;
        }
    }

    public Order(int id, string name, decimal amount)
    {
        OrderId = id;
        CustomerName = name;
        TotalAmount = amount;
        CreationDate = DateTime.Now;
    }

    public virtual void ShowInfo()
    {
        Console.WriteLine($"Заказ #{OrderId} | {CustomerName} | Сумма: {TotalAmount} | Дата: {CreationDate:d}");
    }
}

// Производный класс: Онлайн-заказ (с явной реализацией)
public class OnlineOrder : Order, IOrderManager
{
    public string CustomerEmail { get; set; }

    public OnlineOrder(int id, string name, decimal amount, string email) 
        : base(id, name, amount)
    {
        CustomerEmail = email;
    }

    public override void ShowInfo()
    {
        Console.WriteLine($"[Онлайн] Заказ #{OrderId} | Email: {CustomerEmail} | Сумма: {TotalAmount}");
    }

    public void ProcessOrder()
    {
        Console.WriteLine($"Онлайн-заказ #{OrderId} обработан. Чек отправлен на {CustomerEmail}.");
    }

    // ЯВНАЯ РЕАЛИЗАЦИЯ интерфейса (скрыта от прямого вызова)
    void IOrderManager.CancelOrder()
    {
        Console.WriteLine($"[Система] Отмена онлайн-заказа #{OrderId}. Возврат на карту.");
    }
}

// Производный класс: Физический заказ (обычная реализация)
public class PhysicalOrder : Order, IOrderManager
{
    public string DeliveryAddress { get; set; }

    public PhysicalOrder(int id, string name, decimal amount, string address) 
        : base(id, name, amount)
    {
        DeliveryAddress = address;
    }

    public override void ShowInfo()
    {
        Console.WriteLine($"[Физический] Заказ #{OrderId} | Адрес: {DeliveryAddress} | Сумма: {TotalAmount}");
    }

    public void ProcessOrder()
    {
        Console.WriteLine($"Физический заказ #{OrderId} собран. Курьер выехал на адрес: {DeliveryAddress}.");
    }

    public void CancelOrder()
    {
        Console.WriteLine($"[Система] Отмена физического заказа #{OrderId}. Возврат на склад.");
    }
}

// Сервис с Управлением Зависимостями (Dependency Injection)
public class OrderService
{
    private readonly IOrderManager _orderManager;

    // Внедрение зависимости через конструктор
    public OrderService(IOrderManager orderManager)
    {
        _orderManager = orderManager;
    }

    public void ExecuteOrder()
    {
        Console.WriteLine("--- Начало работы сервиса ---");
        _orderManager.ProcessOrder();
    }

    public void Cancel()
    {
        _orderManager.CancelOrder();
        Console.WriteLine("--- Завершение работы сервиса ---\n");
    }
}

// ==========================================
// ДЕМОНСТРАЦИЯ
// ==========================================

// Работа с онлайн-заказом
var online = new OnlineOrder(101, "Иван", 5000, "ivan@mail.ru");
online.ShowInfo();

// Внедряем онлайн-заказ в сервис
var onlineService = new OrderService(online);
onlineService.ExecuteOrder();
onlineService.Cancel(); // Вызовет явную реализацию

// Работа с физическим заказом
var physical = new PhysicalOrder(202, "Мария", 3000, "ул. Пушкина, д. 10");
physical.ShowInfo();

// Внедряем физический заказ в сервис
var physicalService = new OrderService(physical);
physicalService.ExecuteOrder();
physicalService.Cancel(); // Вызовет обычную реализацию
```
