```
using System;
using System.Collections.Generic;

public interface ISupplier
{
    string GetName();
    decimal GetCost();
    string SupplierType { get; set; }
}

public interface IEmployee
{
    string GetFullName();
    decimal GetSalary();
    string Position { get; set; }
}

public interface IEnterprise
{
    void AddSupplier(ISupplier supplier);
    void AddEmployee(IEmployee employee);
    List<ISupplier> GetSuppliers();
    decimal CalculateTotalExpenses();
}

public class Supplier : ISupplier
{
    public string SupplierType { get; set; }
    private string _name;
    private decimal _cost;

    public Supplier(string name, decimal cost, string type)
    {
        _name = name;
        _cost = cost;
        SupplierType = type;
    }

    public string GetName() => _name;
    public decimal GetCost() => _cost;
}

public class Employee : IEmployee
{
    public string Position { get; set; }
    private string _fullName;
    private decimal _salary;

    public Employee(string fullName, decimal salary, string position)
    {
        _fullName = fullName;
        _salary = salary;
        Position = position;
    }

    public string GetFullName() => _fullName;
    public decimal GetSalary() => _salary;
}

public class Enterprise : IEnterprise
{
    private List<ISupplier> _suppliers = new List<ISupplier>();
    private List<IEmployee> _employees = new List<IEmployee>();

    public void AddSupplier(ISupplier supplier) => _suppliers.Add(supplier);
    public void AddEmployee(IEmployee employee) => _employees.Add(employee);
    public List<ISupplier> GetSuppliers() => _suppliers;

    public decimal CalculateTotalExpenses()
    {
        decimal total = 0;
        foreach (var s in _suppliers) total += s.GetCost();
        foreach (var e in _employees) total += e.GetSalary();
        return total;
    }

    
    public void PrintReport()
    {
        Console.WriteLine("Список поставщиков:");
        for (int i = 0; i < _suppliers.Count; i++)
            Console.WriteLine($"{i + 1}. {_suppliers[i].GetName()} – Стоимость услуг: {_suppliers[i].GetCost()}");

        Console.WriteLine("\nсписок сотрудников:");
        for (int i = 0; i < _employees.Count; i++)
            Console.WriteLine($"{i + 1}. {_employees[i].GetFullName()} – Должность: {_employees[i].Position} – Зарплата: {_employees[i].GetSalary()}");

        Console.WriteLine($"\nОбщие расходы: {CalculateTotalExpenses()}");
    }
}


Console.WriteLine("=== Работа предприятия ===\n");

Enterprise myEnterprise = new Enterprise();

myEnterprise.AddSupplier(new Supplier("Логистический центр", 100000, "Логистика"));
myEnterprise.AddSupplier(new Supplier("Поставщик офисной мебели", 50000, "Оборудование"));

myEnterprise.AddEmployee(new Employee("Андрей Джекович", 70000, "Менеджер"));
myEnterprise.AddEmployee(new Employee("Настя Савельева", 95000, "Разработчик"));


myEnterprise.PrintReport();
```
