<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Order в C#, который будет представлять информацию о
заказах товаров или услуг. На основе этого класса разработать 2-3 производных
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из
классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

<h2 style="color:DodgerBlue">Реализация:</h2>

```
using System;
using System.Collections.Generic;

// Товар
class Item
{
    public string Name { get; set; }
    public double Price { get; set; }

    public Item(string name, double price)
    {
        Name = name;
        Price = price;
    }
}

// Интерфейсы для множественного наследования
interface IDelivery
{
    void Deliver();
}

interface IDiscount
{
    double GetDiscount();
}

// Базовый класс
class Order
{
    public int OrderId { get; set; }
    public DateTime CreationDate { get; set; }
    public double TotalAmount { get; set; }

    public string CustomerName { get; set; }
    public string Status { get; set; }

    protected List<Item> items = new List<Item>();

    public Order(int id, string customer)
    {
        OrderId = id;
        CustomerName = customer;
        CreationDate = DateTime.Now;
        Status = "Создан";
    }

    public virtual void AddItem(Item item)
    {
        items.Add(item);
        CalculateTotal();
    }

    public virtual void RemoveItem(Item item)
    {
        items.Remove(item);
        CalculateTotal();
    }

    public virtual void CalculateTotal()
    {
        TotalAmount = 0;

        foreach (Item item in items)
            TotalAmount += item.Price;
    }

    public virtual void ShowInfo()
    {
        Console.WriteLine($"Заказ №{OrderId}");
        Console.WriteLine($"Клиент: {CustomerName}");
        Console.WriteLine($"Сумма: {TotalAmount}");
        Console.WriteLine($"Статус: {Status}");
    }
}

// Онлайн-заказ
class OnlineOrder : Order, IDelivery
{
    public string CustomerEmail { get; set; }
    public string DeliveryMethod { get; set; }

    public OnlineOrder(int id, string customer, string email)
        : base(id, customer)
    {
        CustomerEmail = email;
        DeliveryMethod = "Курьер";
    }

    public override void AddItem(Item item)
    {
        base.AddItem(item);
        Console.WriteLine($"Товар {item.Name} добавлен. Доставка: {DeliveryMethod}");
    }

    public void Deliver()
    {
        Console.WriteLine("Онлайн-заказ отправлен.");
    }

    public override void ShowInfo()
    {
        base.ShowInfo();
        Console.WriteLine($"Email: {CustomerEmail}");
    }
}

// Физический заказ
class PhysicalOrder : Order
{
    public string DeliveryAddress { get; set; }
    public string Courier { get; set; }

    public PhysicalOrder(int id, string customer, string address)
        : base(id, customer)
    {
        DeliveryAddress = address;
        Courier = "Алексей";
    }

    public override void RemoveItem(Item item)
    {
        base.RemoveItem(item);
        Console.WriteLine($"Товар {item.Name} возвращен.");
    }

    public override void ShowInfo()
    {
        base.ShowInfo();
        Console.WriteLine($"Адрес: {DeliveryAddress}");
    }
}

// Специализированный заказ
// Многоуровневое наследование:
// Order -> OnlineOrder -> SpecializedOrder
class SpecializedOrder : OnlineOrder, IDiscount
{
    public string SpecialConditions { get; set; }
    public double Discount { get; set; }

    public SpecializedOrder(int id, string customer, string email)
        : base(id, customer, email)
    {
        SpecialConditions = "Постоянный клиент";
        Discount = 10;
    }

    public override void CalculateTotal()
    {
        base.CalculateTotal();
        TotalAmount -= TotalAmount * Discount / 100;
    }

    public double GetDiscount()
    {
        return Discount;
    }

    public override void ShowInfo()
    {
        base.ShowInfo();
        Console.WriteLine($"Условия: {SpecialConditions}");
        Console.WriteLine($"Скидка: {Discount}%");
    }
}


// ===========================
// ПРОВЕРКА
// ===========================

Item laptop = new Item("Ноутбук", 1000);
Item mouse = new Item("Мышь", 50);

OnlineOrder online = new OnlineOrder(
    1,
    "Иван",
    "ivan@mail.com"
);

online.AddItem(laptop);
online.AddItem(mouse);
online.ShowInfo();
online.Deliver();

Console.WriteLine();

PhysicalOrder physical = new PhysicalOrder(
    2,
    "Петр",
    "ул. Ленина, 10"
);

physical.AddItem(mouse);
physical.RemoveItem(mouse);
physical.ShowInfo();

Console.WriteLine();

SpecializedOrder special = new SpecializedOrder(
    3,
    "Анна",
    "anna@mail.com"
);

special.AddItem(laptop);
special.ShowInfo();

Console.WriteLine();
Console.WriteLine("=== ПОЛИМОРФИЗМ ===");

Order[] orders = { online, physical, special };

foreach (Order order in orders)
{
    order.ShowInfo();
    Console.WriteLine();
}
```
