```
using System;
using System.Collections.Generic;
using System.Linq;

public delegate void OrderStatusHandler(string message);

public abstract class Order
{
    private decimal _totalAmount;

    public int OrderId { get; set; }
    public DateTime CreationDate { get; set; }
    public string CustomerName { get; set; }
    public string Status { get; protected set; } = "Новый";

    // Событие для отслеживания изменений
    public event OrderStatusHandler StatusChanged;

    public decimal TotalAmount
    {
        get => _totalAmount;
        set
        {
            if (value < 0) throw new ArgumentException("Сумма не может быть отрицательной");
            _totalAmount = value;
        }
    }

    public Order(int id, string name, decimal amount)
    {
        OrderId = id;
        CustomerName = name;
        TotalAmount = amount;
        CreationDate = DateTime.Now;
    }

    public void UpdateStatus(string newStatus)
    {
        Status = newStatus;
        StatusChanged?.Invoke($"Заказ #{OrderId}: статус изменен на '{newStatus}'");
    }

    public virtual void ShowInfo()
    {
        Console.WriteLine($"[Заказ] #{OrderId} | Клиент: {CustomerName} | Сумма: {TotalAmount} | Статус: {Status}");
    }
}

public class OnlineOrder : Order
{
    public string CustomerEmail { get; set; }
    public OnlineOrder(int id, string name, decimal amount, string email) : base(id, name, amount) { CustomerEmail = email; }
    public override void ShowInfo() => Console.WriteLine($"[Онлайн] #{OrderId} | Email: {CustomerEmail} | Сумма: {TotalAmount} | Статус: {Status}");
}

public class PhysicalOrder : Order
{
    public string DeliveryAddress { get; set; }
    public PhysicalOrder(int id, string name, decimal amount, string address) : base(id, name, amount) { DeliveryAddress = address; }
    public override void ShowInfo() => Console.WriteLine($"[Физический] #{OrderId} | Адрес: {DeliveryAddress} | Сумма: {TotalAmount} | Статус: {Status}");
}

// ==========================================
// ДЕМОНСТРАЦИЯ
// ==========================================
Console.WriteLine("=== 1. Коллекции List и Dictionary ===");
List<Order> orderList = new List<Order>
{
    new OnlineOrder(101, "Иван", 5000, "ivan@mail.ru"),
    new PhysicalOrder(202, "Мария", 3000, "ул. Пушкина, 10")
};

Dictionary<int, Order> orderDict = new Dictionary<int, Order>();
foreach (var o in orderList) orderDict.Add(o.OrderId, o);

foreach (var o in orderList) o.ShowInfo();

Console.WriteLine("\nПоиск по ID (словарь):");
if (orderDict.TryGetValue(202, out Order foundOrder)) foundOrder.ShowInfo();

Console.WriteLine("\n=== 2. Делегаты ===");
OrderStatusHandler displayHandler = (msg) => Console.WriteLine($"[Делегат] {msg}");
displayHandler("Обработка заказов завершена.");

Console.WriteLine("\n=== 3. События ===");
orderList[0].StatusChanged += (msg) => Console.WriteLine($"[Событие] {msg}");
orderList[0].StatusChanged += (msg) => Console.WriteLine("[Email-уведомление] Клиенту отправлено письмо.");

orderList[0].UpdateStatus("Оплачен");
orderList[1].UpdateStatus("Доставлен");

Console.WriteLine("\n=== 4. Лямбда-выражения (фильтрация) ===");
var onlineOrders = orderList.Where(o => o is OnlineOrder);
foreach (var o in onlineOrders) o.ShowInfo();
```
